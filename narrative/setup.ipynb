# Flood Analysis Setup: Cheyenne River Flood Event

This notebook sets up a reproducible workflow for analyzing the 2019 flood event on the Cheyenne River near Wasta, South Dakota. The analysis focuses on a time series of streamflow measurements to assess flood frequency and return periods.

## Importance of Time-Series Data for Flood Analysis

Time-series data, collected regularly over time at the same location, is essential for flood analysis. To estimate flood frequency and return periods (e.g., how often a flood of a certain size might occur), we need multiple years of observations. This allows us to model historical patterns and assess the probability of extreme events, helping to inform forecasting and engineering standards.

```python
# Import Required Libraries
# Following PEP-8, imports are in alphabetical order
import earthpy
import geopandas as gpd
import os
import pandas as pd
import plotly.express as px
```

```python
# Declare Key Workflow Variables
# These parameters make the workflow reusable and customizable
analysis_id = "stars"
site_name = "Cheyenne River near Wasta"
year_of_interest = 2019
project_title = "Cheyenne River Flood Frequency"
project_dirname = "flood-cheyenne-data"
```

```python
# Create Project Folder and Download Sample Dataset
# Using earthpy to manage project-specific data directories
project = earthpy.Project(title=project_title, dirname=project_dirname)
data_dir = project.get_data()

# Display the download location
print("Project directory:", project.project_dir)
print("Downloaded data directory:", data_dir)
```

```python
# Verify the Download
# List directory contents to confirm the expected data file is present
print("Directory contents:")
for item in sorted(os.listdir(project.project_dir)):
    print("-", item)

# Confirm the specific file exists
expected_file = "cheyenne_streamflow_1934_2024.csv"
if expected_file in os.listdir(data_dir):
    print(f"\nSuccess: {expected_file} is present in the data directory.")
else:
    print(f"\nWarning: {expected_file} not found. Check the download.")
```
